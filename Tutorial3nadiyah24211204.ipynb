```python

n, p, m = 2, 3, 4
Anp = np.random.rand(n, p)
Apm = np.random.rand(p, m)

DOT = np.dot(Anp, Apm)

print("Matrix 1 :", Anp)
print("\nMatrix 2 :", Apm)
print("\nDot product:", DOT)
```
